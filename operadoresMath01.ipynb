## **Operadores Matemáticos em Python**


```python
# soma em python: mais (+)
1 + 1
```

```python
# subtração em python: menos (-)
2 - 1
```

```python
# multiplicação em python: asterístico (*)
2 * 2
```

```python
# divisão em python: barra (/)
2 / 2
```

```python
# divisão inteira: barra dupla (//)
10 // 3
```

```python
# resto da divisão: porcentagem (%)
10 % 3
```

```python
# potenciação: duplo asterístico (**)
2 ** 3
```

```python
# raíz quadrada: duplo asterístico multiplicado por meio (** 0.5)  
81 ** (1/2)
81 ** 0.5 
```

## **Operadores Booleanos**

```python
# atribuição: igual (=)
cinco = 5
```

```python
# Comparação de igualdade: duplo igual
5 == 3
```

```python
# comparação de diferença: exclamação e igual ( != )
5 != 3
```

```python
# comparação de maioridade/igualdade: maior e igual ( >= )
5 >= 3
# comparação de maioridade: maior (>)
5 > 3
```

```python
# comparação de menoridade/igualdade: menor e igual ( <= )
2 <= 5
# comparação de menoridade: menor (<)
2 < 5
```
