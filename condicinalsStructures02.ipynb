## **Condicionals Structures** *(Control Structures)*

#### IF, ELIF e ELSE

```python
temperatura = float(input("Digite a temperatura em Graus Celsius: "))
if temperatura > 30: 
    print("Está quente!")
elif temperatura < 15:
    print("Está frio!")
else:
    print("Clima está agradável")

# print("Esse print é independente do if")
```

```python
temperatura = float(input("Digite a temperatura em Graus Celsius: "))
if temperatura > 28: 
    print("Está quente!")
elif temperatura >= 15 and temperatura <= 28: 
    print("Clima está agradável")
else:
    if temperatura < 0: 
        print("Está muito frio, congelante!")
    else:
        print("Clima está Frio!!!")
```
