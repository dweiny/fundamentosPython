# **Ties of Repetitions**

### *WHILE TIE*

```python
# WHILE TIE
count = 0
while count <=5:
    print(f"The value of count is {count}")
    count +=1

print("  --Final of while--")
```

### *USE OF CONTINUE*

```python
# USE OF CONTINUE
count = 0
while count <=5:
    print(f"The value of counter is {count}")
    count +=1
    continue # ignore the under code
    print("  -Test of continue-")

print("  --Final of while--")
```

### *USE OF BREAK*

#### *"Duo While"*

```python
# USE OF BREAK
import time

count = 0
while True:
    if count <=10:
        print(f"The value of counter is {count}")
        count +=1
    else:
        break
    print("Here not execut because of break")
print("Loading...")
time.sleep(3)

print("I'm outside of while")
```

### *For Tie*

```python
# FOR TIE
for i in range(10): # traverses a interval
    if i > 5:
        print(i)    
```

```python
names = ["Joao", "Carol", "Pedro", "Ana"]
for name in names:
    print(name)
```

```python
name = "Joao"
for leter in name:
    print(leter)
```
