# ***Saúde e Bem Estar***

```python
print("Bem vindo ao projeto saúde e bem estar")
peso = float(input("Em quilogramas, digite seu peso atual: "))
altura = float(input("Em metros, digite sua altura: "))
IMC = peso/(altura ** 2)
if IMC <18.5:
    print("Você está abaixo do peso!")
elif IMC >18.5 and IMC <24.9:
    print("Você está dentro do peso ideal!")
elif IMC >25.0 and IMC <29.9:
    print("Você está com excesso de peso!")
elif IMC >30.0 and IMC <34.9:
    print("Você está com OBESIDADE CLASSE I!")
elif IMC >35 and IMC <39.9:
    print("Você está com OBESIDADE CLASSE II!!")
elif IMC >=40:
    print("Você está com OBESIDADE CLASSE III!!!")
else:
    print("Erro")
```

# **Cálculo de Renda**

```python
valorH = float(input("Em reais, digite o valor de sua hora trabalhada: "))
horasT = int(input("Insira a quantidade de horas trabalhadas: "))
rendaB = valorH * horasT
impostoR = rendaB * 0.11
INSS = rendaB * 0.08
Sind = rendaB * 0.05
RL = rendaB - impostoR - INSS - Sind
print(f"Seu salário bruto é de: R${rendaB}.")
print(f"Desconto ao Imposto de Renda: R${impostoR}.")
print(f"Desconto ao INSS: R${INSS}.")
print(f"Desconto ao Sindicato: R${Sind}.")
print(f"Portanto, seu salário líquido é de R${RL}.")
```
