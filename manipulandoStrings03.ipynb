## **Acesso à caractéres individualmente**

```python
minhaString = "hello world"

# Acessar caracteres individualmente
minhaString[0] # H
minhaString[1] # e
minhaString[2] # l
minhaString[3] # l
minhaString[4] # o
```

## **Concatenação de Strings**

```python
# concatenar strings
outraString = " world!"
novaString = minhaString + outraString
print(novaString)
```

## **Repetição de Strings**

```python
# repetir strings
repetida = minhaString * 4
print(repetida)
```

## **Fatiamento de Strings**


```python
# fatiar Strings
minhaString[1:4] # do índice 1 ao 4
minhaString[:] # quando o índice antecessor dos dois pontos não definido, basea-se no índice 0
minhaString[:] # string completa (índices não definidos)
minhaString[::] # de quantas em quantas casas
minhaString[-1] # execução contrária
minhaString[::-1] # execução contrária da String

# dois pontos (:) equivale á entre
```

## **Funções de lista**

```python
print(type(minhaString)) # tipo de código
print(dir(minhaString)) # funções compatíveis
len(minhaString) # tamanho da String
len("pneumoultramicrospiossilicovulcanicaniotico") # tamanho da String
minhaString.upper() # tudo para cima, tudo maiúsculo
minhaString.lower() # tudo para baixo, tudo minúsculo
minhaString.capitalize() # primeira palavra em maiúsculo
novaString.title() # primeiras letras em maiúsculo
minhaString.strip() # remove espaços em branco
minhaString.split() # divisão das palavras # (indice de corte)
minhaString.replace("l", "x") # troca de caractéres
# duas trocas= duplo replace
minhaString.find("h") # revela o índice do primeiro caractére encontrado
minhaString.count("l") # revela quantidade de caractéres
minhaString.startswith("h") # T/F- afirmações sobre o início da String
minhaString.endswith("ld") # T/F- afirmações sobre o término da String
minhaString.sort # ordenação crescente
minhaString.reverse # ordenação reversa


print("ell" in minhaString) # T/F- averiguação se a String possui ou não tais caractéres
```
